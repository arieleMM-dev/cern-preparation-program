```python
import numpy as np
import matplotlib.pyplot as plt

# Generamos un dataset de prueba de pacientes rápido y lo guardamos
# 10 pacientes que seran las filas y 5 días de registro que serán las columnas
datos_simulados = np.random.randint(0,20, size=(10,5))
np.savetxt('pacientes_inflamacion.csv',datos_simulados,delimiter=',')

# 1. Carga de datos
datos = np.loadtxt(fname='pacientes_inflamacion.csv',delimiter=',')
print("Datos cargados correctamente.")
```

```python
# 2. Exploración
print("Tipo de estructura:",type(datos))
print("Dimensiones del dataset (pacientes,días):",datos.shape)
print("\nPrimeros 3 registros (pacientes 0 al 2):\n",datos[:3,:])
```

```python
# 3. Análisis

# Operación 1: Promedio general por cada paciente
promedio_por_paciente = np.mean(datos,axis=1)
print("Promedio de inflamación por paciente:\n",promedio_por_paciente)

# Operación 2: Pico máximo registrado en cada día
maximo_por_dia = np.max(datos,axis=0)
print("\nMáxima inflamación registrada por día:\n",maximo_por_dia)
```

```python
# 4. Visualización
fig = plt.figure(figsize=(10.0,3.0))
axes1 = fig.add_subplot(1,2,1)
axes2 = fig.add_subplot(1,2,2)
axes1.set_ylabel('Promedio')
axes1.set_xlabel('Paciente')
axes1.set_title('Promedio de inflamación por paciente')
axes1.plot(promedio_por_paciente,marker='o',color='blue')
axes2.set_ylabel('Máximo')
axes2.set_xlabel('Día')
axes2.set_title('Pico máximo de inflamación diario')
axes2.plot(maximo_por_dia, marker='s',color='red')
fig.tight_layout()
plt.show()
```

### 5.Interpretación de los Resultados

* **Exploración inicial:** Comprobé que los datos se importaron bien como una matriz bidimensional de NumPy con 10 filas y 5 columnas siendo las filas los pacientes y las columnas los días de 
registro.
* **Análisis de promedios:** Al ver el primer gráfico, se ve que el promedio de inflamación varía bastante entre cada persona.Por ejemplo, el paciente 2 tuvo el promedio más alto con 14.4, mientras que el paciente 9 tuvo el más bajo con 5.2.
* **Análisis de picos máximos:** En el segundo gráfico, al evaluar la inflamación máxima diaria, me pareció interesante ver que el primer día el pico fue de 16, pero a partir del segundo día subió y se estancó en un máximo exacto de 19, manteniéndose totalmente plano el resto de los días.Esto refleja claramente cómo los datos generados de forma aleatoria alcanzaron rápidamente un límite alto constante en las columnas, aunque con datos reales pueden marcar una tendencia diferente, lo cuál seria interesante de analizar.El tema de datos me interesa bastante, ya que muchas veces situaciones, acciones o comportamientos que tenemos diariamente pueden traducirse a tendencias o conductas para estudiarlas.
